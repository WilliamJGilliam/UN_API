```python
url = 'https://api.worldbank.org/v2/country/USA;CAN/indicator/NY.GDP.PCAP.PP.KD'

params = {
    'format': 'json',
    'date': '2000:2021',
    'per_page': 50,
    'page': 1
}
```

```python
all_data = []

response = requests.get(url, params=params)
data = response.json()

total_pages = data[0]['pages']

for page in range(1, total_pages + 1):
    params['page'] = page

    response = requests.get(url, params=params)
    data = response.json()

    all_data.extend(data[1])
```

```python
all_data
```

```python
indicator_url = 'https://api.worldbank.org/v2/indicator'

indicator_params = {
    'format': 'json',
    'per_page': 50,
    'page': 1
}
```

```python
all_indicators = []

response = requests.get(indicator_url, params=indicator_params)
data = response.json()

total_pages = data[0]['pages']

for page in range(1, total_pages + 1):
    indicator_params['page'] = page

    response = requests.get(indicator_url, params=indicator_params)
    data = response.json()

    all_indicators.extend(data[1])
```

```python
indicators_df = pd.json_normalize(all_indicators)
```

```python
education_indicator = indicators_df[
    indicators_df['name'].str.contains(
        'Public Expenditure on Education',
        case=False,
        na=False
    )
]
```
